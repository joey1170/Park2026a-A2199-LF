# 2. Convert to csv file

```python
import numpy as np
import pandas as pd
import os

# Path to your .npz file
npz_fname = '/Users/jonginpark/A2199_LF/DATA/A2199/z_A2199_Hwang/z_DATA/z_a2199phot21_5DR9_hshwang.npz'

# Load the .npz file
npz_data = np.load(npz_fname, allow_pickle=True)

# Convert to native byte order and handle byte strings
data_dict = {}
for key in npz_data.files:
    arr = npz_data[key]

    # Handle byte order
    if arr.dtype.byteorder == '>' or (arr.dtype.byteorder == '=' and np.little_endian is False):
        arr = arr.byteswap().newbyteorder()

    # Convert byte strings to regular strings
    if arr.dtype.kind in {'S', 'O'}:  # S = bytes, O = object (might contain bytes)
        arr = np.array([x.decode('utf-8') if isinstance(x, bytes) else x for x in arr])

    data_dict[key] = arr

# Build DataFrame and save to CSV
df = pd.DataFrame(data_dict)
csv_fname = os.path.splitext(npz_fname)[0] + '.csv'
df.to_csv(csv_fname, index=False)

print(f"Saved cleaned DataFrame to CSV at: {csv_fname}")
```
